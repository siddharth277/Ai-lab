## **1.1 `for` loop demonstartion**
### **1.1.1 Example 1 – Print first 5 diamond prices**

```python
for i in range(5):
  print(f'Diamond, {i+1}, Price : {data['price'][i]}')
  print("Diamond:", i+1, "Price:", data['price'][i])
```

### **1.1.2 Example 2 – Display carat & cut of first 10 diamonds**

```python
for i in range(10):
  carat = data["carat"][i]
  cut = data["cut"][i]
  print("Diamond:", i+1, "Carat:", carat, "Cut:", cut)
```

### **1.1.3 Example 3 – Count how many diamonds have carat > 1**

```python
count = 0

for i in data['carat']:
  if i > 1:
    count +=1

print("Count of diamonds with carat > 1:", count)
```

## **1.2 `while` loop demonstration**

### **1.2.1 Example 4 – Print diamonds until price exceeds 20,000**

```python
i = 0
while i < len(data) and data['price'][i] <=20000:
  print("Diamond:", i+1, "Price:", data['price'])
  i +=1
```

### **1.2.2 Example 5 – Sum prices of first 15 diamonds**

```python
i = 0
total_price = 0

while i < 15:
  total_price += data['price'][i]
  i += 1
print("Total price of first 15 diamonds :", total_price)
```

```python
scores = np.array([88, 92, 79, 95, 85])

min_index = np.argmin(scores)

print("Scores:", scores)
print("Index of minimum value:", min_index)
print("Minimum value:", scores[min_index])
```

```python
## 2-D array
marks = np.array([[85, 90, 88],
                  [78, 92, 80]])

min_pos = np.argmin(marks)
row, col = np.unravel_index(min_pos, marks.shape)

print("Minimum value:", marks[row, col])
print("Located at row", row, "column", col)
```

### **3.10 np.where(): Selecting Values That Satisfy a Condition**

```python
# np.where()
marks = np.array([35, 68, 72, 41, 90, 28])

passed = np.where(marks >= 40)

print("Indices of students who passed:", passed)
print("Marks of students who passed:", marks[passed])
```
