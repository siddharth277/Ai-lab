```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
```

```python
df_train = pd.read_csv("Data_for_Lab06/Activity_Train.csv")
df_test = pd.read_csv("Data_for_Lab06/Activity_Test.csv")
df_val = pd.read_csv("Data_for_Lab06/Activity_Validation.csv")
```

```python
X_train = df_train.drop('Activity', axis=1)
y_train = df_train['Activity']
```

```python
X_test = df_test.drop('Activity', axis=1)
y_test = df_test['Activity']
```

```python
X_val = df_val.drop('Activity', axis=1)
y_val = df_val['Activity']
```

```python
logistic_reg = LogisticRegression ( max_iter =1000)

logistic_reg . fit ( X_train , y_train )

y_test_pred = logistic_reg . predict ( X_test )
weights = logistic_reg . coef_
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))
acc_lr_orig = accuracy_score(y_test, y_test_pred)
print("Accuracy:",acc_lr_orig)
print("Precision (Micro):", precision_score(y_test, y_test_pred, average="micro"))
print("Recall (Micro):", recall_score(y_test, y_test_pred, average="micro"))
print("F1-score (Micro):", f1_score(y_test, y_test_pred, average="micro"))

print("Precision (Macro):", precision_score(y_test, y_test_pred, average="macro"))
print("Recall (Macro):", recall_score(y_test, y_test_pred, average="macro"))
print("F1-score (Macro):", f1_score(y_test, y_test_pred, average="macro"))
```

```python
logistic_reg = LogisticRegression ( max_iter =1000)

logistic_reg . fit ( X_train , y_train )

y_val_pred = logistic_reg . predict ( X_val )
weights = logistic_reg . coef_
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_pred))

print("Accuracy:", accuracy_score(y_val, y_val_pred))
print("Precision (Micro):", precision_score(y_val, y_val_pred, average="micro"))
print("Recall (Micro):", recall_score(y_val, y_val_pred, average="micro"))
print("F1-score (Micro):", f1_score(y_val, y_val_pred, average="micro"))

print("Precision (Macro):", precision_score(y_val, y_val_pred, average="macro"))
print("Recall (Macro):", recall_score(y_val, y_val_pred, average="macro"))
print("F1-score (Macro):", f1_score(y_val, y_val_pred, average="macro"))
```

```python
#svm poly kernal 
```

```python
best_degree, best_val_acc = None, 0
for degree in [2, 3, 4, 5]:
  poly_svm = SVC(kernel="poly", degree=degree,C=1.0, gamma="scale")

  poly_svm.fit(X_train, y_train)
  val_acc = poly_svm.score(X_val, y_val)
  print(f"Degree={degree}, Validation Accuracy={val_acc:.4f}")
  if val_acc > best_val_acc:
    best_val_acc = val_acc
    best_degree = degree
print(f"\nBest Degree: {best_degree}")
best_poly_svm = SVC(kernel="poly", degree=best_degree,C=1.0, gamma="scale")

best_poly_svm.fit(X_train, y_train)
y_test_pred = best_poly_svm.predict(X_test)
y_val_pred = best_poly_svm.predict(X_val)
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))
acc_poly_orig= accuracy_score(y_test, y_test_pred)
print("Accuracy:",acc_poly_orig)
print("Precision (Micro):", precision_score(y_test, y_test_pred, average="micro"))
print("Recall (Micro):", recall_score(y_test, y_test_pred, average="micro"))
print("F1-score (Micro):", f1_score(y_test, y_test_pred, average="micro"))

print("Precision (Macro):", precision_score(y_test, y_test_pred, average="macro"))
print("Recall (Macro):", recall_score(y_test, y_test_pred, average="macro"))
print("F1-score (Macro):", f1_score(y_test, y_test_pred, average="macro"))
```
