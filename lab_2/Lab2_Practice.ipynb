# **Plot a histogram with a Kernel Density Estimation (KDE) curve for a selected numerical attribute and mark its mean, median, and mean ±2.7σ values on the plot.**

### **Using matplotlib**

```python
cols = ['Unit_Price', 'Revenue', 'Profit']

plt.figure(figsize=(12,8))

for i, col in enumerate(cols, 1):

    d = data_num[col].dropna()

    mu = d.mean()
    median = d.median()
    sigma = d.std(ddof=1)

    ucl = mu + 2.7 * sigma
    lcl = mu - 2.7 * sigma

    plt.subplot(2,2,i)

    # Histogram
    plt.hist(d, bins=30, density=True, alpha=0.6, edgecolor='black')

    # KDE
    d.plot(kind='kde', linewidth=2)

    # Reference lines
    plt.axvline(mu, linestyle='--', linewidth=2, label='Mean')
    plt.axvline(median, linestyle='-.', linewidth=2, label='Median')
    plt.axvline(ucl, linestyle=':', linewidth=2, color='red', label='+2.7σ')
    plt.axvline(lcl, linestyle=':', linewidth=2, color='red', label='-2.7σ')

    # Numeric labels
    ymax = plt.ylim()[1]
    plt.text(mu, ymax*0.90, f"{mu:.2f}", ha='center')
    plt.text(median, ymax*0.80, f"{median:.2f}", ha='center')
    plt.text(ucl, ymax*0.70, f"+2.7σ = {ucl:.2f}", ha='center', color='red')
    plt.text(lcl, ymax*0.70, f"-2.7σ = {lcl:.2f}", ha='center', color='red')

    plt.title(col)
    plt.xlabel(col)
    plt.ylabel("Density")
    plt.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

### **Using Seaborn**

```python
cols = ['Unit_Price', 'Revenue', 'Profit']

plt.figure(figsize=(12,8))

for i, col in enumerate(cols, 1):

    d = data_num[col].dropna()

    mu = d.mean()
    median = d.median()
    sigma = d.std(ddof=1)

    ucl = mu + 2.7 * sigma
    lcl = mu - 2.7 * sigma

    plt.subplot(2,2,i)

    # Histogram + KDE
    sns.histplot(d, bins=30, stat="density", kde=True)

    # Reference lines
    plt.axvline(mu, linestyle='--', linewidth=2, label='Mean')
    plt.axvline(median, linestyle='-.', linewidth=2, label='Median')
    plt.axvline(ucl, linestyle=':', linewidth=2, color='red', label='+2.7σ')
    plt.axvline(lcl, linestyle=':', linewidth=2, color='red', label='-2.7σ')

    # Numeric labels
    ymax = plt.ylim()[1]
    plt.text(mu, ymax*0.90, f"{mu:.2f}", ha='center')
    plt.text(median, ymax*0.80, f"{median:.2f}", ha='center')
    plt.text(ucl, ymax*0.70, f"+2.7σ = {ucl:.2f}", ha='center', color='red')
    plt.text(lcl, ymax*0.70, f"-2.7σ = {lcl:.2f}", ha='center', color='red')

    plt.title(col)
    plt.xlabel(col)
    plt.ylabel("Density")
    plt.legend(fontsize=8)

plt.tight_layout()
plt.show()
```
